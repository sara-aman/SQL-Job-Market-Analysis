#### Skill Category Analysis --- Machine Learning Job Market

This script loads the exported SQL results (CSV) and generates a bar chart showing the average salary per skill category for Machine Learning roles.

Environment: Python 3.9 (Anaconda)

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load CSV file
df = pd.read_csv(r"C:\Users\Sara\Desktop\Self_Learing\SQL_YouTube\visualization\Skill_Category_Analysis.csv")
```

```python
# Display the data 
plt.figure(figsize=(10,7))
plt.bar(df['skill_category'], df['avg_salary'], color='skyblue')

plt.title("Average Salary by Skill Category (ML Roles)")
plt.xlabel("Skill Category")
plt.ylabel("Average Salary ($)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
